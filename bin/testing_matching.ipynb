## Merging them together

### Local plans

```python
merged = pd.merge(old_df[['reference', 'entity', 'organisations']], my_df[['reference', 'local-planning-authorities']], on='reference', how='outer', indicator=True)
```

```python
merged['_merge'].value_counts()
```

### Local plan documents

```python
old_df_doc.columns
```

```python
merged_doc = pd.merge(old_df_doc[['reference', 'entity', 'organisation']], my_df_doc['reference'], on='reference', how='outer', indicator=True)
```

```python
merged_doc['_merge'].value_counts()
```

## Match based on matching references and share entity number

```python
merged_match = merged.rename(columns={'local-planning-authorities':'my_work', 'organisations':'old_work'}).loc[merged['_merge'] =='both'][['reference', 'entity']]

# Drop pre-existing entity column and map across matched entities
my_df_dropped = my_df.drop(columns=['entity'])
my_df_dropped['entity']=my_df_dropped['reference'].map(
    merged_match.set_index('reference')['entity']
)
```

```python
## Files to give to Swati

my_df_dropped.loc[~my_df_dropped['entity'].isna()].to_csv('/Users/sianteesdale/Documents/GitHub/local-plan-extractor/dataset/testing/local-plan.csv', index=False)

my_df_dropped_doc.loc[~my_df_dropped_doc['entity'].isna()].to_csv('/Users/sianteesdale/Documents/GitHub/local-plan-extractor/dataset/testing/local-plan-document.csv', index=False)
```
